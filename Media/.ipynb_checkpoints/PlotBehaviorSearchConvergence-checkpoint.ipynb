## Visualizing values by percentile slabs (GREAT TO REPEAT THESE FOR THE FULL SEARCH):


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from scipy.stats import gaussian_kde

def plot_percentile_slabs_heatmap(dataframe, search_name, names_of_params_of_interest):
    """
    Heatmap showing the density of parameter values across fitness percentile slabs
    """
    
    os.makedirs('analysis/plots/BehaviorSearch/percentileSlabs', exist_ok=True)
    
    n_slabs = 20
    percentiles = np.linspace(0, 100, n_slabs + 1)
    
    for param_name in names_of_params_of_interest:
        param_values = dataframe[param_name]
        fitness_values = dataframe['fitness']
        
        # Create bins for parameter values
        n_param_bins = 30
        param_bins = np.linspace(param_values.min(), param_values.max(), n_param_bins + 1)
        param_centers = (param_bins[:-1] + param_bins[1:]) / 2
        
        # Create 2D histogram: slabs x parameter bins
        hist_2d = np.zeros((n_slabs, n_param_bins))
        
        for i in range(n_slabs):
            lower_percentile = percentiles[i]
            upper_percentile = percentiles[i + 1]
            
            lower_bound = fitness_values.quantile(lower_percentile / 100)
            upper_bound = fitness_values.quantile(upper_percentile / 100)
            
            if i == 0:
                mask = fitness_values <= upper_bound
            elif i == n_slabs - 1:
                mask = fitness_values > lower_bound
            else:
                mask = (fitness_values > lower_bound) & (fitness_values <= upper_bound)
            
            slab_values = param_values[mask]
            
            # Histogram of parameter values in this slab
            hist, _ = np.histogram(slab_values, bins=param_bins)
            hist_2d[i, :] = hist
        
        # Normalize each row to show relative distribution within each slab
        row_sums = hist_2d.sum(axis=1, keepdims=True)
        row_sums[row_sums == 0] = 1  # Avoid division by zero
        hist_2d_normalized = hist_2d / row_sums
        
        # Create heatmap
        fig, ax = plt.subplots(figsize=(14, 6))
        
        im = ax.imshow(hist_2d_normalized, aspect='auto', cmap='viridis', 
                      extent=[param_bins[0], param_bins[-1], n_slabs, 0])
        
        ax.set_xlabel(f'{param_name} Value', fontsize=12)
        ax.set_ylabel('Fitness Percentile Slab (0=Best, 19=Worst)', fontsize=12)
        ax.set_title(f'{param_name} Distribution Across Fitness Percentiles - {search_name}', fontsize=14)
        
        # Set y-ticks
        y_ticks = np.arange(0, n_slabs, 2)
        y_labels = [f'{percentiles[i]:.0f}-{percentiles[i+1]:.0f}%' for i in y_ticks]
        ax.set_yticks(y_ticks)
        ax.set_yticklabels(y_labels)
        
        # Colorbar
        cbar = plt.colorbar(im, ax=ax, pad=0.02)
        cbar.set_label('Relative Density within Slab', fontsize=10)
        
        plt.tight_layout()
        
        # Save
        clean_name = param_name.replace('*', '').replace('?', '').replace('[', '').replace(']', '')
        plt.savefig(f'analysis/plots/BehaviorSearch/percentileSlabs/percentile_slabs_heatmap_{search_name}_{clean_name}.png', 
                   dpi=300, bbox_inches='tight')
        plt.close()

# Run for both searches
plot_percentile_slabs_heatmap(objFunctionHistory_Tsize60, 'Tsize60', names_of_params_of_interest)
plot_percentile_slabs_heatmap(objFunctionHistory_stepwise[-1], 'Stepwise', names_of_params_of_interest)
```

```python
def plot_percentile_slabs_summary(dataframe, search_name, names_of_params_of_interest):
    """
    Summary plot showing median, quartiles, and min/max of parameter values across slabs
    """
    
    os.makedirs('analysis/plots/BehaviorSearch/percentileSlabs', exist_ok=True)
    
    n_slabs = 20
    percentiles = np.linspace(0, 100, n_slabs + 1)
    
    for param_name in names_of_params_of_interest:
        param_values = dataframe[param_name]
        fitness_values = dataframe['fitness']
        
        slab_stats = {
            'median': [],
            'q1': [],
            'q3': [],
            'min': [],
            'max': [],
            'count': []
        }
        slab_labels = []
        
        for i in range(n_slabs):
            lower_percentile = percentiles[i]
            upper_percentile = percentiles[i + 1]
            
            lower_bound = fitness_values.quantile(lower_percentile / 100)
            upper_bound = fitness_values.quantile(upper_percentile / 100)
            
            if i == 0:
                mask = fitness_values <= upper_bound
            elif i == n_slabs - 1:
                mask = fitness_values > lower_bound
            else:
                mask = (fitness_values > lower_bound) & (fitness_values <= upper_bound)
            
            slab_values = param_values[mask]
            
            if len(slab_values) > 0:
                slab_stats['median'].append(np.median(slab_values))
                slab_stats['q1'].append(np.percentile(slab_values, 25))
                slab_stats['q3'].append(np.percentile(slab_values, 75))
                slab_stats['min'].append(np.min(slab_values))
                slab_stats['max'].append(np.max(slab_values))
                slab_stats['count'].append(len(slab_values))
            else:
                slab_stats['median'].append(np.nan)
                slab_stats['q1'].append(np.nan)
                slab_stats['q3'].append(np.nan)
                slab_stats['min'].append(np.nan)
                slab_stats['max'].append(np.nan)
                slab_stats['count'].append(0)
            
            slab_labels.append(f'{lower_percentile:.0f}-{upper_percentile:.0f}%')
        
        # Create figure
        fig, ax = plt.subplots(figsize=(14, 6))
        
        x_positions = np.arange(n_slabs)
        
        # Plot median line
        ax.plot(x_positions, slab_stats['median'], 'b-', linewidth=2, label='Median')
        
        # Plot IQR ribbon
        ax.fill_between(x_positions, slab_stats['q1'], slab_stats['q3'], 
                        alpha=0.3, color='blue', label='IQR (25-75%)')
        
        # Plot min-max range
        ax.fill_between(x_positions, slab_stats['min'], slab_stats['max'], 
                        alpha=0.1, color='blue', label='Min-Max Range')
        
        ax.set_xlabel('Fitness Percentile Slab (0=Best fitness)', fontsize=12)
        ax.set_ylabel(f'{param_name} Value', fontsize=12)
        ax.set_title(f'{param_name} Summary Across Fitness Percentiles - {search_name}', fontsize=14)
        ax.set_xticks(x_positions[::2])
        ax.set_xticklabels(slab_labels[::2], rotation=45, ha='right')
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        
        clean_name = param_name.replace('*', '').replace('?', '').replace('[', '').replace(']', '')
        plt.savefig(f'analysis/plots/BehaviorSearch/percentileSlabs/percentile_slabs_summary_{search_name}_{clean_name}.svg', 
                   dpi=300, bbox_inches='tight')
        plt.savefig(f'analysis/plots/BehaviorSearch/percentileSlabs/percentile_slabs_summary_{search_name}_{clean_name}.png', 
                   dpi=300, bbox_inches='tight')
        plt.close()

# Run for both searches
plot_percentile_slabs_summary(objFunctionHistory_Tsize60, 'Tsize60', names_of_params_of_interest)
plot_percentile_slabs_summary(objFunctionHistory_stepwise[-1], 'Stepwise', names_of_params_of_interest)
```
